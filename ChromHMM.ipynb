```python
# -----------------------------
# 0️⃣ Import bibliotek
# -----------------------------
import os
import glob
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ttest_rel
from statsmodels.stats.multitest import multipletests
import numpy as np

# -----------------------------
# 1️⃣ Wybrane próbki grupy negative
# -----------------------------
samples_negative = ["Ind04","Ind10","Ind17","Ind34","Ind37","Ind38","Ind43"]

# -----------------------------
# 2️⃣ Znajdź pliki .bed dla tych próbek
# -----------------------------
all_files = glob.glob("output/*.bed")
selected_files = []

for f in all_files:
    filename = os.path.basename(f)
    for s in samples_negative:
        if filename.startswith(s) and ("d0" in filename or "d1" in filename):
            selected_files.append(f)

print("Liczba plików w grupie negative:", len(selected_files))
print(selected_files[:10])

# -----------------------------
# 3️⃣ Wczytaj dane i przygotuj dataframe
# -----------------------------
data = []

for f in selected_files:
    df = pd.read_csv(f, sep="\t", header=None)
    
    state = df.iloc[:,-1]  # ostatnia kolumna
    sample = os.path.basename(f).split("_")[0]
    day = "d0" if "d0" in f else "d1"
    
    tmp = pd.DataFrame({
        "state": state,
        "sample": sample,
        "day": day
    })
    
    data.append(tmp)

data = pd.concat(data, ignore_index=True)

# -----------------------------
# 4️⃣ Mapowanie stanów E1-E15 na nazwy epigenetyczne
# -----------------------------
state_map = {
"E1": "Active TSS","E2": "Flanking Active TSS","E3": "Transcription (5' and 3')",
"E4": "Strong transcription","E5": "Weak transcription","E6": "Genic enhancer",
"E7": "Enhancer","E8": "ZNF genes & repeats","E9": "Heterochromatin",
"E10": "Bivalent/Poised TSS","E11": "Flanking Bivalent TSS/Enh","E12": "Bivalent enhancer",
"E13": "Repressed PolyComb","E14": "Weak Repressed PolyComb","E15": "Quiescent / Low signal"
}

data["state"] = data["state"].map(state_map)

# -----------------------------
# 5️⃣ Wyliczenie frakcji stanów per sample/day
# -----------------------------
summary = (
    data
    .groupby(["sample","day","state"])
    .size()
    .reset_index(name="count")
)

summary["fraction"] = summary.groupby(["sample","day"])["count"].transform(lambda x: x/x.sum())

# -----------------------------
# 6️⃣ Pivot do tabeli: sample x state (d0 i d1)
# -----------------------------
df_wide = summary.pivot_table(index="sample", columns=["day","state"], values="fraction", fill_value=0)

# -----------------------------
# 7️⃣ Paired t-test per state + FDR
# -----------------------------
states = summary["state"].unique()
results = []

for state in states:
    try:
        d0 = df_wide[("d0", state)]
        d1 = df_wide[("d1", state)]
    except KeyError:
        continue
    
    t_stat, p_val = ttest_rel(d0, d1)
    
    results.append({
        "state": state,
        "t_stat": t_stat,
        "p_value": p_val
    })

results_df = pd.DataFrame(results)

# FDR correction
reject, pvals_corrected, _, _ = multipletests(results_df["p_value"], alpha=0.05, method="fdr_bh")
results_df["p_value_FDR"] = pvals_corrected
results_df["significant"] = reject

results_df = results_df.sort_values("p_value_FDR")
print("Paired t-test per state (negative group) with FDR correction:")
print(results_df)

# -----------------------------
# 8️⃣ Barplot - wszystkie próbki
# -----------------------------
plt.figure(figsize=(12,6))
sns.barplot(
    data=summary,
    x="sample",
    y="fraction",
    hue="state"
)
plt.xticks(rotation=45)
plt.ylabel("Fraction")
plt.title("Epigenetic Roadmap annotations - Negative group")
plt.legend(bbox_to_anchor=(1.05,1))
plt.tight_layout()
plt.show()

# -----------------------------
# 9️⃣ Barplot z rozdzieleniem d0/d1
# -----------------------------
g = sns.catplot(
    data=summary,
    x="sample",
    y="fraction",
    hue="state",
    col="day",
    kind="bar",
    height=6,
    aspect=1.2
)
for ax in g.axes.flat:
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45)
g.fig.suptitle("Fraction of epigenetic states per sample (Negative group, d0 vs d1)", y=1.05)
plt.show()

# -----------------------------
# 🔹 10️⃣ Heatmapa różnic d1 - d0 (tylko stany istotne po FDR)
# -----------------------------
diff = df_wide["d1"] - df_wide["d0"]

# Wybierz tylko kolumny odpowiadające istotnym stanom po FDR
sig_states = results_df[results_df["significant"]]["state"].values
diff_sig = diff[sig_states]

plt.figure(figsize=(14,6))
sns.heatmap(diff_sig, annot=True, fmt=".2f", cmap="RdBu", center=0)
plt.title("Difference in fraction of epigenetic states (d1 - d0), FDR significant states")
plt.ylabel("Sample")
plt.xlabel("Epigenetic state")
plt.show()
```

```python
import os, glob, re, time
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.stats import ttest_rel
from statsmodels.stats.multitest import multipletests

plt.rcParams['svg.fonttype'] = 'none'

print(">>> Working directory:", os.getcwd())

state_map = {
    "E1": "Active TSS", "E2": "Flanking Active TSS",
    "E3": "Transcription (5' and 3')", "E4": "Strong transcription",
    "E5": "Weak transcription", "E6": "Genic enhancer",
    "E7": "Enhancer", "E8": "ZNF genes & repeats",
    "E9": "Heterochromatin", "E10": "Bivalent/Poised TSS",
    "E11": "Flanking Bivalent TSS/Enh", "E12": "Bivalent enhancer",
    "E13": "Repressed PolyComb", "E14": "Weak Repressed PolyComb",
    "E15": "Quiescent / Low signal",
}
state_order = list(state_map.values())
state_colors = {
    "Active TSS": "#D62839", "Flanking Active TSS": "#E76F51",
    "Transcription (5' and 3')": "#A7C957", "Strong transcription": "#588157",
    "Weak transcription": "#3A5A40", "Genic enhancer": "#E9C46A",
    "Enhancer": "#F4D35E", "ZNF genes & repeats": "#4ECDC4",
    "Heterochromatin": "#9D8DF1", "Bivalent/Poised TSS": "#E07A5F",
    "Flanking Bivalent TSS/Enh": "#F2B5A0", "Bivalent enhancer": "#B08968",
    "Repressed PolyComb": "#6C757D", "Weak Repressed PolyComb": "#ADB5BD",
    "Quiescent / Low signal": "#E9ECEF",
}
group_colors = {"Positive": "firebrick", "Moderate": "forestgreen", "Negative": "steelblue"}

def run_chromhmm_pipeline(samples, group_name, bed_dir="output"):
    files = glob.glob(os.path.join(bed_dir, "*.bed"))
    selected_files = [f for f in files if any(s in f for s in samples)]
    if not selected_files:
        raise ValueError(f"[{group_name}] No files matched — check bed_dir/sample IDs.")
    data_list = []
    for f in selected_files:
        df = pd.read_csv(f, sep="\t", header=None)
        state = df.iloc[:, -1]
        sample = os.path.basename(f).split("_")[0]
        match = re.search(r'\d+(d\d+)', f)
        day = match.group(1) if match else "unknown"
        data_list.append(pd.DataFrame({"state": state, "sample": sample, "day": day}))
    data = pd.concat(data_list, ignore_index=True)
    data["state"] = data["state"].map(state_map).fillna(data["state"])
    summary = data.groupby(["sample", "day", "state"]).size().reset_index(name="count")
    summary["fraction"] = summary.groupby(["sample", "day"])["count"].transform(lambda x: x / x.sum())
    df_wide = summary.pivot_table(index="sample", columns=["day", "state"], values="fraction", fill_value=0)
    states = summary["state"].unique()
    results = []
    for state in states:
        if all((d, state) in df_wide.columns for d in ["d0", "d1"]):
            t, p = ttest_rel(df_wide[("d0", state)], df_wide[("d1", state)])
            results.append({"state": state, "t_stat": t, "p_value": p})
    results_df = pd.DataFrame(results)
    reject, p_fdr, _, _ = multipletests(results_df["p_value"], alpha=0.05, method="fdr_bh")
    results_df["p_value_FDR"] = p_fdr
    results_df["significant"] = reject
    diff = df_wide["d1"] - df_wide["d0"]
    print(f"[{group_name}] samples matched: {summary['sample'].nunique()} | files: {len(selected_files)}")
    return {"summary": summary, "diff": diff, "results_df": results_df}

try:
    samples_moderate = ["Ind02","Ind03","Ind05","Ind11","Ind12","Ind13","Ind15","Ind16",
                         "Ind18","Ind23","Ind28","Ind29","Ind35","Ind39","Ind40","Ind42","Ind09","Ind45"]
    samples_positive = ["Ind06","Ind07","Ind08","Ind22","Ind24","Ind25","Ind26","Ind27",
                         "Ind31","Ind33","Ind36","Ind41","Ind44"]
    samples_negative = ["Ind04","Ind10","Ind17","Ind34","Ind37","Ind38","Ind43"]

    results_store = {}
    for group, samples in [("Moderate", samples_moderate),
                            ("Positive", samples_positive),
                            ("Negative", samples_negative)]:
        results_store[group] = run_chromhmm_pipeline(samples, group)

    TSS_LABEL = "Active TSS (+ Flanking)"
    OTHER_LABEL = "Other states"
    merge_map = {"Active TSS": TSS_LABEL, "Flanking Active TSS": TSS_LABEL, "Enhancer": "Enhancer"}
    for s in state_order:
        if s not in merge_map:
            merge_map[s] = OTHER_LABEL
    grouped_order = [TSS_LABEL, "Enhancer", OTHER_LABEL]
    grouped_colors = [state_colors["Active TSS"], state_colors["Enhancer"], "#B0BEC5"]

    print(">>> Grupowanie stanów:")
    for new_cat in grouped_order:
        print(f"    {new_cat}: {[s for s, v in merge_map.items() if v == new_cat]}")

    def aggregate_to_grouped(fraction_series):
        tmp = fraction_series.reindex(state_order).fillna(0)
        tmp.index = tmp.index.map(merge_map)
        return tmp.groupby(level=0).sum().reindex(grouped_order).fillna(0)

    fig, axes = plt.subplots(1, 3, figsize=(15, 6))
    for ax, group in zip(axes, group_colors):
        summ = results_store[group]["summary"]
        mean_by_day = summ.groupby(["day", "state"])["fraction"].mean().reset_index()
        d0_full = mean_by_day[mean_by_day["day"] == "d0"].set_index("state")["fraction"].reindex(state_order).fillna(0)
        d1_full = mean_by_day[mean_by_day["day"] == "d1"].set_index("state")["fraction"].reindex(state_order).fillna(0)
        d0 = aggregate_to_grouped(d0_full)
        d1 = aggregate_to_grouped(d1_full)
        ax.pie(d1, radius=1.0, colors=grouped_colors, wedgeprops=dict(width=0.35, edgecolor="white", linewidth=1))
        ax.pie(d0, radius=0.65, colors=grouped_colors, wedgeprops=dict(width=0.35, edgecolor="white", linewidth=1))
        ax.set_title(group, color=group_colors[group], fontweight="bold", fontsize=13)

        out_path = os.path.join(os.getcwd(), f"chromHMM_pie_{group.lower()}.svg")
        fig_i, ax_i = plt.subplots(figsize=(5, 5))
        ax_i.pie(d1, radius=1.0, colors=grouped_colors, wedgeprops=dict(width=0.35, edgecolor="white", linewidth=1))
        ax_i.pie(d0, radius=0.65, colors=grouped_colors, wedgeprops=dict(width=0.35, edgecolor="white", linewidth=1))
        ax_i.set_title(group, color=group_colors[group], fontweight="bold", fontsize=13)
        fig_i.tight_layout()
        fig_i.savefig(out_path, bbox_inches="tight", transparent=True)
        plt.close(fig_i)
        print(f">>> Zapisano: {out_path}  (mtime: {time.ctime(os.path.getmtime(out_path))})")

    handles = [mpatches.Patch(color=c, label=s) for c, s in zip(grouped_colors, grouped_order)]
    fig.legend(handles=handles, title="Chromatin state", bbox_to_anchor=(1.05, 0.5), loc="center left")
    fig.text(0.5, 0.02, "Inner ring = Day 0   •   Outer ring = Day 1", ha="center", fontsize=10, style="italic")
    fig.suptitle("ChromHMM chromatin state composition per group (d0 vs d1)", y=1.03)
    fig.tight_layout()

    combined_path = os.path.join(os.getcwd(), "chromHMM_pies_by_group.svg")
    fig.savefig(combined_path, bbox_inches="tight")
    print(f">>> Zapisano: {combined_path}  (mtime: {time.ctime(os.path.getmtime(combined_path))})")
    plt.show()

except Exception as e:
    print("!!! BŁĄD:", repr(e))
    import traceback
    traceback.print_exc()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import shapiro

def significance_stars(p):
    if p < 0.001: return "***"
    elif p < 0.01: return "**"
    elif p < 0.05: return "*"
    else: return "ns"

group_colors = {"Positive": "firebrick", "Moderate": "forestgreen", "Negative": "steelblue"}

fig, axes = plt.subplots(1, 3, figsize=(11, 4.5), sharey=True)
shapiro_results = {}

for ax, group in zip(axes, ["Positive", "Moderate", "Negative"]):
    summ = results_store[group]["summary"]
    res = results_store[group]["results_df"]
    color = group_colors[group]

    enh = (summ[summ["state"] == "Enhancer"]
           .pivot_table(index="sample", columns="day", values="fraction", fill_value=0)
           .reindex(columns=["d0", "d1"]))
    d0_vals = enh["d0"].values * 100  # jako % genomu
    d1_vals = enh["d1"].values * 100
    diff_vals = d1_vals - d0_vals

    row = res[res["state"] == "Enhancer"]
    p_fdr = row.iloc[0]["p_value_FDR"] if not row.empty else np.nan
    stars = significance_stars(p_fdr) if not np.isnan(p_fdr) else "n/a"

    # Shapiro-Wilk na różnicach Day1 - Day0
    if len(diff_vals) >= 3:
        sw_stat, sw_p = shapiro(diff_vals)
    else:
        sw_stat, sw_p = np.nan, np.nan
    shapiro_results[group] = {"n": len(diff_vals), "W": sw_stat, "p": sw_p}

    # paired lines, lekko przezroczyste
    for y0, y1 in zip(d0_vals, d1_vals):
        ax.plot([0, 1], [y0, y1], color=color, alpha=0.45, linewidth=1.1, zorder=2)

    # kropki indywidualne, bez czarnej obwódki
    ax.scatter([0]*len(d0_vals), d0_vals, color=color, alpha=0.85, s=38, edgecolor="none", zorder=3)
    ax.scatter([1]*len(d1_vals), d1_vals, color=color, alpha=0.85, s=38, edgecolor="none", zorder=3)

    # średnia grupy: czarny diament + przerywana linia
    ax.plot([0, 1], [d0_vals.mean(), d1_vals.mean()], color="black",
            linestyle="--", linewidth=2, zorder=4)
    ax.scatter([0, 1], [d0_vals.mean(), d1_vals.mean()], color="black",
               marker="D", s=55, zorder=5)

    ax.set_xlim(-0.3, 1.3)
    ax.set_xticks([0, 1])
    ax.set_xticklabels(["Day 0", "Day 1"])
    ax.set_title(group, fontweight="bold", fontsize=12, color="black", pad=30)

    ax.text(0.5, 1.09, f"FDR p = {p_fdr:.3g}  {stars}" if not np.isnan(p_fdr) else "n/a",
            transform=ax.transAxes, ha="center", va="bottom", fontsize=8.5, color="dimgray")
    ax.text(0.5, 1.02, f"Shapiro-Wilk p = {sw_p:.3g}" if not np.isnan(sw_p) else "Shapiro-Wilk: n/a",
            transform=ax.transAxes, ha="center", va="bottom", fontsize=7.5, color="gray", style="italic")

    # pełna ramka panelu (jak theme_bw)
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_color("black")
        spine.set_linewidth(0.8)
    ax.grid(False)

# tick marki tylko na lewym panelu
for ax in axes[1:]:
    ax.tick_params(axis="y", which="both", length=0)

axes[0].set_ylabel("Enhancer fraction of genome (%)")
axes[0].set_ylim(bottom=0)

plt.tight_layout()
plt.savefig("enhancer_significance_styled.svg", bbox_inches="tight")
plt.show()

# ---- Podsumowanie do Methods ----
print("\nShapiro-Wilk test on Day1-Day0 Enhancer differences:")
for group, r in shapiro_results.items():
    verdict = "normal (paired t-test appropriate)" if r["p"] >= 0.05 else "NOT normal (consider Wilcoxon)"
    print(f"  {group}: n={r['n']}, W={r['W']:.3f}, p={r['p']:.3g} -> {verdict}")
```

```python
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

group = "Positive"
summ = results_store[group]["summary"]
res = results_store[group]["results_df"].copy()

enh = (summ[summ["state"] == "Enhancer"]
       .pivot_table(index="sample", columns="day", values="fraction", fill_value=0)
       .reindex(columns=["d0", "d1"]))
d0_vals = enh["d0"].values
d1_vals = enh["d1"].values

w_stat, w_p_raw = wilcoxon(d0_vals, d1_vals)
print(f"Wilcoxon signed-rank (Positive, Enhancer): W={w_stat:.3f}, raw p={w_p_raw:.4g}")

# oznacz który test był użyty per stan
if "test_used" not in res.columns:
    res["test_used"] = "paired t-test"
res.loc[res["state"] == "Enhancer", "p_value"] = w_p_raw
res.loc[res["state"] == "Enhancer", "test_used"] = "Wilcoxon"

# przelicz FDR na nowo dla WSZYSTKICH 15 stanów w Positive (z podmienionym p dla Enhancera)
reject, p_fdr_new, _, _ = multipletests(res["p_value"], alpha=0.05, method="fdr_bh")
res["p_value_FDR"] = p_fdr_new
res["significant"] = reject

results_store["Positive"]["results_df"] = res  # nadpisz w results_store

print(res.sort_values("p_value_FDR")[["state", "test_used", "p_value", "p_value_FDR", "significant"]])
```
